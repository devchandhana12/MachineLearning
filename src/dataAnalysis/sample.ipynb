```python
import pandas as pd
df = pd.read_csv("telecom_customer_churn.csv")
df[(df["total_charges"] == " ") & (df["tenure"] == 0)].replace(" ", "0.0", inplace=True)

df["churn"].unique()

df["customer_id"].duplicated().sum()

df["total_charges"] = pd.to_numeric(df["total_charges"], errors="coerce")

df[df["total_charges"] < 0]

df["churn"].value_counts(normalize=True) * 100

subset = df[(df["tenure"] < 12) & (df["contract"] == "Month-to-month") & (df["internet_service"] == "Fiber optic")  & (df["tech_support"] == "No") & (df["online_security"] == "No")]

subset["churn"].value_counts()





```
