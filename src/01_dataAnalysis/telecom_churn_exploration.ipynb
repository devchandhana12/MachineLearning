# Exploring telecom customer churn

Start by loading the data and checking its size and columns. This notebook is a small practice in understanding a dataset before fitting a model.

```python
from pathlib import Path
import pandas as pd

data_path = Path('src/01_dataAnalysis/telecom_customer_churn.csv')
if not data_path.exists():
    data_path = Path('telecom_customer_churn.csv')

df = pd.read_csv(data_path)
print(f'{len(df):,} rows and {len(df.columns)} columns')
df.head()
```

## Check data quality

Clean blank charge values before converting the column to numbers. Then inspect missing values and duplicate customer IDs.

```python
blank_charges = df['total_charges'].astype(str).str.strip().eq('')
zero_tenure = df['tenure'].eq(0)
df.loc[blank_charges & zero_tenure, 'total_charges'] = '0.0'
df['total_charges'] = pd.to_numeric(df['total_charges'], errors='coerce')

print('Duplicate customer IDs:', df['customer_id'].duplicated().sum())
display(df.isna().sum().sort_values(ascending=False).head(10))
df.loc[df['total_charges'] < 0]
```

## Explore churn patterns

Compare the overall target balance, then examine one customer segment. These are observations to investigate, not proof that any feature causes churn.

```python
print('Churn labels:', df['churn'].unique())
display((df['churn'].value_counts(normalize=True) * 100).rename('percent'))

segment = df[
    (df['tenure'] < 12)
    & (df['contract'] == 'Month-to-month')
    & (df['internet_service'] == 'Fiber optic')
    & (df['tech_support'] == 'No')
    & (df['online_security'] == 'No')
]
segment['churn'].value_counts()
```
